### LangChainとエージェント

LangChainは、LLMアプリを開発するためのフレームワーク.
チェーンという仕組みで、ようやく、分析、というタスクを順に
処理していける。少し複雑なアプリ作成に。

LangGraphというエージェントを構築できる。モデルが自分で判断して
ツールを選んだりDBにアクセスしたりして、動的にタスクを進められる。

外部ツールと組み合わせて使える。
ウェブAPIにアクセス、ベクトルDBやSQLDBの検索、カスタム計算ツール

様々な言語モデルを統一されたインターフェイスで扱える
プロンプトもテンプレートで効率よく作れる

例えば作れるもの：

　　レポート自動作成、集計とまとめ文章化、プランの自動作成・・・

◆モジュールのインポート　準備

```python
# 必要なモジュールをインポート
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage

# 環境変数の読み込み
load_dotenv("../.env")
os.environ['OPENAI_API_KEY'] = os.environ['API_KEY']

# モデル名
MODEL_NAME = "gpt-4o-mini"
```

◆言語モデルの呼び出し

```python
# モデルの作成
chat_model = ChatOpenAI(model_name=MODEL_NAME)
#ここでは「gpt-4o-mini」＝Chatモデルを使っている
# 例えば一問一答型のCompletionsモデルなら OpenAI() を使用


# 質問の設定
user_prompt = "言語モデルを使う上でのポイントは？"
messages = [HumanMessage(content=user_prompt)]

# 言語モデルの呼出
response = chat_model.invoke(messages)

# 結果を表示
print(response.content)
```

◆質問を設定
OpenAI APIでは、質問の種類を role で指定したが、LangChainでは質問の種類ごとのクラスがある。
　role	LangChain	content
　system	SystemMessage	前提や役割
　user	HumanMessage	ユーザーからの質問
　assistant	AIMessage	言語モデルからの回答

```python
# 質問の設定
user_prompt = "言語モデルを使う上でのポイントは？"
messages = [HumanMessage(content=user_prompt)]
```

◆言語モデルの呼び出しと、その結果表示

```python
# 言語モデルの呼出
response = chat_model.invoke(messages)  #呼び出しにはinvoke

# 結果を表示。戻り値は ”AIMessage 型”。そんなのあるのね
print(response.content)
```

◆システムプロンプトと、パラメータの設定
（ネコの設定にしてる）

```python
# モデルの作成
chat_model = ChatOpenAI(
    model_name=MODEL_NAME,
    max_tokens=300,
    temperature=1.2)

# 質問の設定
system_prompt = "あなたは猫です。にゃーと答えます。回答文は、もし長ければ区切って改行を入れて。"
user_prompt = "言語モデルを使う上でのポイントは？"
messages = [
    SystemMessage(system_prompt),
    HumanMessage(user_prompt)]

# 言語モデルの呼出と結果の表示（ストリーミング）
for chunk in chat_model.stream(messages):  # streamメソッドを使う
    print(chunk.content, end="", flush=True)
```

◆プロンプトテンプレート

テンプレートとパラメータを組み合わせる

```python
from langchain_core.prompts import ChatPromptTemplate
#Chatモデル用のプロンプトテンプレートをインポート

system_template = "あなたは{input_language}から{output_language}に翻訳する優秀な翻訳家です。"
human_template = "{text}"
#パラメータとなる部分は｛｝に入れて。

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", system_template),
    ("human", human_template),
])
# "system" でシステムプロンプト SystemMessage 、
# "human" でユーザープロンプト HumanMessage を定義。

messages = chat_prompt.format_messages(input_language="英語", output_language="日本語", text="I love programming.")

# 作成されたプロンプトは、通常のプロンプトと同じように使える
messages
```

```python
# モデルの作成
chat_model = ChatOpenAI(model_name=MODEL_NAME)

# 言語モデルの呼出　invoke
response = chat_model.invoke(messages)

# 結果を表示
print(response.content)
```



#### Output parser　とは

言語モデルからの出力を、CSVやJSON、指定されたクラスなどに変換する機能。
ここではカンマ区切りの文字列を、CommaSeparatedListOutputParser で扱う

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser

# モデルの作成
chat_model = ChatOpenAI(model_name=MODEL_NAME)

# 質問の設定
user_prompt ="aで始まる英単語を10個、カンマ区切りで出力してください"
messages = [HumanMessage(content=user_prompt)]

# 言語モデルの呼出
response = chat_model.invoke(messages)

# Output Parserの作成　この時点では文字列。リスト化への準備
output_parser = CommaSeparatedListOutputParser()

# Output parserで変換　ここでリストへ
word_list = output_parser.parse(response.content)
print(type(word_list))
print(word_list)
```

#### チェーン

処理と処理を | でつなぐ LCEL（LangChain Expression Language）
という記法で書く。

```python
#まずはチェーンを使わない書き方。
# 処理と処理の間でデータを受け渡す必要。


# プロンプトテンプレートの作成
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "あなたは{animal}らしく、語尾に{voice}などと付けて答えます。"),
    ("human", "{question}をする上でのポイントは？"),
])

messages = chat_prompt.format_messages(animal="犬", voice="ワン！", question="英語学習")

# モデルの作成
chat_model = ChatOpenAI(model_name=MODEL_NAME)

# 言語モデルの呼出
response = chat_model.invoke(messages)

# 結果を表示
print(response.content)
```

```python
#　次にチェーンを使った書き方
# 

from langchain_core.prompts import ChatPromptTemplate

# プロンプトテンプレートの作成
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "あなたは{animal}らしく、語尾に{voice}などと付けて答えます。"),
    ("human", "{question}をする上でのポイントは？"),
])

# モデルの作成
chat_model = ChatOpenAI(model_name=MODEL_NAME)

# チェーンの作成
chain = chat_prompt | chat_model
#ここでチェーンを作っている。前の処理の戻り値と、後の処理の引数は
# 一致している必要あり。

# チェーンの実行
response = chain.invoke({"animal": "犬", "voice": "ワン！", "question": "英語学習"})

# 結果を表示
print(response.content)
```
