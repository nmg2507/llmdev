#### APIリクエスト

```python
# 必要なモジュールをインポート
import os
from dotenv import load_dotenv
from openai import OpenAI
from pprint import pprint

# 環境変数の取得 (APIキーでGPT環境をつなぐ。
# 今のファイルのある場所から見て、1つ上のフォルダにある 
# .env ファイル　を指している。
load_dotenv("../.env")


# openAI　APIくらいあんとを生成
client = OpenAI(api_key = os.environ['API_KEY'])

# モデル名
MODEL_NAME = "gpt-4o-mini"

# ここまでは実行しても出力はなし。準備のみ。
# 必須パラメータである model と messages だけを指定した
# シンプルなプログラムをもとに、APIリクエストとレスポンスの
# 基本について確認。

# メッセージの設定
message = "言語モデルを使う上でのポイントは"

# APIへリクエスト
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": message},
    ]
)

# レスポンスの表示
pprint(vars(response))
```


role｜	content

system｜	前提や役割

user｜	ユーザーからの質問

assistant｜	言語モデルからの回答

tool｜	ツールの呼び出し

下記のようにこうすると、会話の内容が保持されます。

messages=[ {"role": "user", "content": "こんにちは"}, 

　　　　{"role": "assistant", "content": "こんにちは！"}, 

　　　　{"role": "user", "content": "元気？"}, ]


```python
# 言語モデルからの回答を表示
print(response.choices[0].message.content.strip())
```

役割や前提の設定

メッセージの role に "system" を指定することで、「ユーザーからの質問」とは別に「役割や前提」を設定できます。

```python
# 役割や前提の設定
role = "あなたはマーケティング分野に精通したビジネスコンサルタントです。企業の成長をサポートするために、効果的なマーケティング戦略を提供します。"

# メッセージの設定
message = "新製品の発売に向けた効果的なマーケティング戦略を教えてください。"

# APIへリクエスト
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "system", "content": role},
        {"role": "user", "content": message},
    ],
)

# 言語モデルからの回答を表示
print(response.choices[0].message.content.strip())
```

#### 各種パラメータの説明　(出力の多様性など)

パラメータで、やりとりを調節する

```python
# 出力の多様性
message = "日本の魅力を紹介して！"

# APIへリクエスト
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": message},
    ],
    temperature=1.4,   # 標準1.0。値を大きくすることで多様性が大きく
    n = 3, # 3回答を得る
    max_completion_tokens = 100 # 出力トークン長を制限
)

# 結果を表示（複数回答を比較）
for choice in response.choices:
    print("-" * 20)
    print(choice.message.content.strip())
```

```python
# 出力の多様性
message = "日本の魅力を紹介して！"

# APIへリクエスト
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": message},
    ],
    presence_penalty=-2.0, # マイナス値で同じ単語を選ばれやすく
    n = 3, # 3回答を得る
    max_completion_tokens = 100 # 出力トークン長を制限
)

# 結果を表示（複数回答を比較）
for choice in response.choices:
    print("-" * 20)
    print(choice.message.content.strip())
```

#### トークンの選択度合い：logit_bias

logit_bias は、指定されたトークンの選択度合い（選ばれやすさ／にくさ）を調整。通常-10～+10の間で調整するが、例えば-100とか大きな数値に指定すると、使わせない、などの制御ができる。

```python
import tiktoken
from tiktoken.core import Encoding

# OpenAI APIの特定のモデルに対応するトークナイザーを取得
encoding: Encoding = tiktoken.encoding_for_model(MODEL_NAME) # o200k_base

# テキストをトークンIDのリストに変換
tokens = encoding.encode("こんにちは！")
tokens_count = len(tokens)

# トークンの長さとトークンIDを表示
print(f"{tokens_count=}")
print(f"{tokens=}")
```

```python
# 得られたトークンIDを選ばれやすくする
# トークンの選択度合い
message = """
AさんとBさんで会話してください。
A:あ、Bさんだ。
B:
"""

# APIへリクエスト
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": message},
    ],
    max_tokens=100, # 使用トークン数を制限
    n=3,
    logit_bias = {95839:8, 3393:8} # ８という強いバイアス
)

# 結果を表示
for choice in response.choices:
    print("-" * 20)
    print(choice.message.content.strip())
```

#### 回答をストリーミングで得る：stream
Web版ChatGPTのように、言語モデルからの回答を、単語ごとに逐次ストリームで得る場合に指定します。

使用ぱらめーた：stream

```python
# メッセージの設定
message = "言語モデルを使う上でのポイントは"

# APIへリクエｒスト
stream = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": message},
    ],
    stream=True,  # ここ！
)

for chunk in stream:
    if chunk.choices:
        if chunk.choices[0].delta.content is not None:
            print(chunk.choices[0].delta.content,end="")
```
